```python
from rag import vector_search, build_prompt, generate_answer, build_index


def ask_assistant(rag, question: str, top_k: int = 3) -> str:
    """Полный RAG-пайплайн: поиск релевантных чанков, сборка промпта, вызов LLM."""
    retrieved = vector_search(rag, question, top_k=top_k)
    prompt = build_prompt(question, retrieved)
    return generate_answer(prompt)
```

```python
# Индексация: запускается один раз за сессию ядра.
# Ячейку ниже можно перезапускать сколько угодно - индекс уже готов.
rag = build_index("docs", show_progress_bar=True)
```

```python
question = "Какой HTTP-метод использует эндпоинт /api/generate?"
print(f"Вопрос: {question}\n")
print(f"Ответ:\n{ask_assistant(rag, question)}")
```

## Тестирование ассистента

```python
test_questions = [
    'Какой HTTP-метод использует эндпоинт /api/generate?',
    'Что такое Modelfile?',
    'Как удалить локально скачанную модель через CLI?',
    'Какая текущая цена подписки на Ollama Cloud?'
]


def debug_ask(raq, question: str, top_k: int = 3):
    """Показывает и найденные чанки, и финальный ответ - удобно для отладки."""
    print(f"Вопрос: {question} \n")

    retrieved = vector_search(raq, question, top_k=top_k)
    print("Найденные чанки:")
    for i, r in enumerate(retrieved, 1):
        print(f"  #{i} [{r['source']}] score={r['score']:.3f}")
        print(f"     {r['text'][:120].strip()}...")
    print()

    prompt = build_prompt(question, retrieved)
    answer = generate_answer(prompt)
    print(f"Ответ модели:\n{answer}\n")


for question in test_questions:
    debug_ask(rag, question)
    print("-" * 70)
    print()
```
